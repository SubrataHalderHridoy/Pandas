```python
import pandas as pd
```

```python
s = pd.Series([1, 2, 3, 4, 5])
print(s)
print(type(s))

# index
print(s[0])
print(s[2])
```

```python
s = pd.Series([12, 13, 14, 15, 16], index=["Hridoy", "Eve", "Adam", "Bob", "Charlie"])
print(s)
print(s.iloc[0])
```

```python
# Homogeneous

s = pd.Series([12, 13, 14, 15, 16, "abc"], index=["Hridoy", "Eve", "Adam", "Bob", "Charlie", 25])
print(s)
print(s.iloc[0])
print(s.index)
```

```python
# Vectorized Operations

s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([12, 22, 32, 43, 52])
print(s1 + s2)
```

```python
# Handle Missing values with NaN

s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([12, 22, 32, 43, 52])
print(s1 + s2)

s1[0] = 100
print(s1)
```

```python
# Mutable values, immutable size

s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([12, 22, 32, 43, 52])
print(s1 + s2)

s1[0] = 100
changed_s1 = s1.drop(0)

print(s1)
print(changed_s1)
```
